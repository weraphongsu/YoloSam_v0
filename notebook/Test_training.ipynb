```python
from ultralytics import YOLO
# Load a pre-trained YOLO model
model = YOLO('D:/SIG/Yolo/pretrain/yolov10m.pt')  
# Train the model
model.train(
    data='D:/SIG/Yolo/notebook/data.yaml',  # Path to your dataset config
    epochs=50,                # Number of epochs
    imgsz=640,                 # Image size for training
    batch=8,                  # Batch size
    lr0=0.01,
    device="0",                   # Use GPU (0), or "cpu" for CPU
    project='D:/SIG/Yolo/training_results',  
    name='train_medium_exp'
)
```
